# Calcul de proba des résultats d'attaques

**Objectif** : On souhaite spécifier deux unités ainsi que des informations situationels et obtenir une probabilité pour le nombre de (wonds, damage, models retirés).

> Note : on utilisera le vocabulaire / mots-clés en anglais.

## Structure d'une attaque

Les [regles de bases](https://www.warhammer-community.com/en-gb/downloads/warhammer-40000/). On y définit la séquence suivante HIT ROLL > WOUND ROLL > ALLOCATE ATTACK > SAVING THROW > INFLICT DAMAGE. Chaque profil d'arme de l'unité attaquante ainsi que ses abilitées et les abilités de l'unité attaque affectent les différentes étapes de la séquence. Afin de correctement séquencer la résolutions de l'attaque on ajoutera l'étape "FIX STATS".

| Etape           | Mot clé d'arme ayant un effet | Stat/abilité ayant un effet |
|-----------------|-------------------------------|---------------------|
| FIX STATS       | Rapid fire - Blast - Anti-keyword | (A) |
| HIT ROLL        | Torent - Lethal hit - Heavy - Sustained hit - Indirect fire | (BS) - (WS) |
| WOUND ROLL      | Lethal hit - Twin linked - Lance - Devastating wound - Sustained hit | (S) - (T) |
| ALLOCATE ATTACK | | |
| SAVING THROW    | Ignore cover - Indirect Fire | (Sv) - Invulnerable save - Benefit of cover |
| INFLICT DAMAGE  | Devastating wound - Melta | Feel no pain |

On ignore les mots clés : Assault - Pistol - Precision - Hazardous - Extra attacks

### Détail des mots clés

| Mot clé | Détail |
|---------|--------|
| Anti-keyword | 
| Blast | Ajoutez 1 à la caractéristique (A) pour chaque groupe de cinq figurines dans l'unité cible (en arrondissant à l'unité inférieure).
| Devastating wound |
| Heavy | Ajoute 1 au hit roll si attaquant n'a pas bougé à ce tour.
| Indirect fire |
| Ignore cover |
| Lance |
| Lethal hit |
| Melta |
| Rapid fire x | Augmente le nombre d'attaques de x si la cible est à demi porté.
| Sustained hit | 
| Torent | 
| Twin linked | Chaque fois qu'une attaque est lancée avec une telle arme, vous pouvez relancer le WOUND ROLL de cette attaque.
